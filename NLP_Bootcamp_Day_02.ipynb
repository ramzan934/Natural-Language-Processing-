Sentiment analysis

```python
from nltk.sentiment import SentimentIntensityAnalyzer

sia = SentimentIntensityAnalyzer()

text = "This product is excellent and I really love it!"

scores = sia.polarity_scores(text)

print("Sentiment scores:", scores)

if scores['compound'] >= 0.05:
    print("Sentiment: Positive")
elif scores['compound'] <= -0.05:
    print("Sentiment: Negative")
else:
    print("Sentiment: Neutral")
```

Text summarization

```python
import nltk
import re
from collections import Counter
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.corpus import stopwords

text = """
Natural Language Processing is a field of artificial intelligence.
It helps computers process human language.
NLP is used in chatbots, translation, and sentiment analysis.
Machine learning helps computers learn patterns from text.
Modern NLP systems also use transformer models.
"""

sentences = sent_tokenize(text)
stop_words = set(stopwords.words('english'))

words = [
    w.lower() for w in word_tokenize(text)
    if w.isalpha() and w.lower() not in stop_words
]

freq = Counter(words)

scores = {}
for sentence in sentences:
    sentence_words = [
        w.lower() for w in word_tokenize(sentence)
        if w.isalpha()
    ]
    scores[sentence] = sum(
        freq[w] for w in sentence_words
    ) / max(len(sentence_words), 1)

summary = sorted(
    sorted(sentences, key=lambda s: scores[s], reverse=True)[:2],
    key=sentences.index
)

print("Summary:", " ".join(summary))
```

Text classification

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import make_pipeline

texts = [
    "The team won the football match",
    "The player scored a goal",
    "The computer uses artificial intelligence",
    "New software improves computer performance",
    "The government announced a new policy",
    "The election campaign started today"
]

labels = [
    "Sports", "Sports",
    "Technology", "Technology",
    "Politics", "Politics"
]

model = make_pipeline(
    TfidfVectorizer(),
    LogisticRegression()
)

model.fit(texts, labels)

new_text = ["The football player scored a goal"]
prediction = model.predict(new_text)

print("Predicted category:", prediction[0])
```

Simple Chatbot

```python
def chatbot():
    print("Bot: Hello! Type 'bye' to exit.")

    while True:
        user = input("You: ").strip().lower()

        if user == "bye":
            print("Bot: Goodbye!")
            break
        elif "hello" in user or "hi" in user:
            print("Bot: Hello! How can I help you?")
        elif "your name" in user:
            print("Bot: I am a simple NLP chatbot.")
        elif "nlp" in user:
            print("Bot: NLP helps computers process human language.")
        elif "how are you" in user:
            print("Bot: I am working well. Thank you!")
        else:
            print("Bot: Sorry, I don't understand that yet.")

chatbot()
```
