```python
# Complete Regular Expressions (Regex) Program for NLP

import re

print("=" * 60)
print("       REGULAR EXPRESSIONS (REGEX) IN NLP")
print("=" * 60)

# --------------------------------------------------
# 1. BASIC REGEX: FIND ALL MATCHES
# --------------------------------------------------
print("\n1. FIND ALL MATCHES")

text = "Ali has 25 books, 10 pens, and 3 notebooks."

print("Numbers:", re.findall(r"\d+", text))
print("Words:", re.findall(r"[A-Za-z]+", text))
print("Capital letters:", re.findall(r"[A-Z]", text))
print("Small letters:", re.findall(r"[a-z]", text))


# --------------------------------------------------
# 2. REGEX SPECIAL CHARACTERS
# --------------------------------------------------
print("\n2. SPECIAL CHARACTERS")

text = "abc abbc abbbc ac"

print("a followed by zero or more b:", re.findall(r"ab*c", text))
print("a followed by one or more b:", re.findall(r"ab+c", text))
print("a followed by zero or one b:", re.findall(r"ab?c", text))
print("Exactly three digits:", re.findall(r"\d{3}", "123 45 678 9"))
print("Two to four digits:", re.findall(r"\d{2,4}", "1 12 123 12345"))


# --------------------------------------------------
# 3. SEARCH: FIRST MATCH ANYWHERE
# --------------------------------------------------
print("\n3. SEARCH")

text = "I am learning NLP in 2026."

result = re.search(r"\d+", text)

if result:
    print("First number:", result.group())
    print("Start position:", result.start())
    print("End position:", result.end())
else:
    print("No match found")


# --------------------------------------------------
# 4. MATCH: CHECK THE BEGINNING
# --------------------------------------------------
print("\n4. MATCH")

print(re.match(r"Hello", "Hello World"))
print(re.match(r"World", "Hello World"))


# --------------------------------------------------
# 5. FULLMATCH: CHECK THE ENTIRE STRING
# --------------------------------------------------
print("\n5. FULLMATCH")

print(bool(re.fullmatch(r"\d{4}", "2026")))  # True
print(bool(re.fullmatch(r"\d{4}", "26")))    # False
print(bool(re.fullmatch(r"\d{4}", "2026a"))) # False


# --------------------------------------------------
# 6. REPLACE TEXT USING SUB
# --------------------------------------------------
print("\n6. TEXT REPLACEMENT")

text = "I love NLP!!! My score is 100."

print("Remove digits:", re.sub(r"\d+", "", text))
print("Remove punctuation:", re.sub(r"[^\w\s]", "", text))
print("Replace NLP:", re.sub(r"NLP", "AI", text))


# --------------------------------------------------
# 7. SPLIT TEXT
# --------------------------------------------------
print("\n7. SPLIT TEXT")

text = "Ali, Ahmed; Sara. Hassan"

print("Split on punctuation:",
      re.split(r"[,;.]\s*", text))

text = "apple   banana\torange"

print("Split on whitespace:",
      re.split(r"\s+", text))


# --------------------------------------------------
# 8. CHARACTER CLASSES
# --------------------------------------------------
print("\n8. CHARACTER CLASSES")

text = "Ali_123 @ Lahore!"

print("Digits:", re.findall(r"\d", text))
print("Non-digits:", re.findall(r"\D+", text))
print("Word characters:", re.findall(r"\w+", text))
print("Non-word characters:", re.findall(r"\W", text))
print("Whitespace:", re.findall(r"\s", text))
print("Non-whitespace:", re.findall(r"\S+", text))


# --------------------------------------------------
# 9. START, END, AND WORD BOUNDARIES
# --------------------------------------------------
print("\n9. BOUNDARIES")

text = "NLP is useful. NLP is powerful."

print("Lines/strings starting with NLP:",
      bool(re.search(r"^NLP", text)))

print("Text ending with powerful:",
      bool(re.search(r"powerful\.$", text)))

print("Whole word 'cat':",
      re.findall(r"\bcat\b", "cat category cats"))


# --------------------------------------------------
# 10. GROUPS
# --------------------------------------------------
print("\n10. GROUPS")

text = "Name: Ali, Age: 20"

result = re.search(r"Name:\s*(\w+),\s*Age:\s*(\d+)", text)

if result:
    print("Full match:", result.group(0))
    print("Name:", result.group(1))
    print("Age:", result.group(2))
    print("All groups:", result.groups())


# --------------------------------------------------
# 11. NAMED GROUPS
# --------------------------------------------------
print("\n11. NAMED GROUPS")

text = "Email: ali@example.com"

result = re.search(
    r"Email:\s*(?P<email>[\w.+-]+@[\w.-]+\.[A-Za-z]{2,})",
    text
)

if result:
    print("Email:", result.group("email"))


# --------------------------------------------------
# 12. ALTERNATION: OR OPERATOR
# --------------------------------------------------
print("\n12. ALTERNATION")

text = "I like Python, Java, and NLP."

print("Find Python or Java:",
      re.findall(r"Python|Java", text))


# --------------------------------------------------
# 13. CASE-INSENSITIVE SEARCH
# --------------------------------------------------
print("\n13. IGNORE CASE")

text = "Python python PYTHON"

print(re.findall(r"python", text, flags=re.IGNORECASE))


# --------------------------------------------------
# 14. EMAIL EXTRACTION
# --------------------------------------------------
print("\n14. EMAIL EXTRACTION")

text = """
Contact ali@example.com or sara123@gmail.com.
Invalid example: ali@com
"""

pattern = r"\b[\w.+-]+@[\w.-]+\.[A-Za-z]{2,}\b"

print("Emails:", re.findall(pattern, text))


# --------------------------------------------------
# 15. PHONE NUMBER EXTRACTION
# --------------------------------------------------
print("\n15. PHONE NUMBER EXTRACTION")

text = "Call 03001234567 or +92 300 1234567."

pattern = r"(?<!\w)(?:\+92[\s-]?|0)3\d{2}[\s-]?\d{7}(?!\w)"

print("Phone numbers:", re.findall(pattern, text))


# --------------------------------------------------
# 16. URL EXTRACTION
# --------------------------------------------------
print("\n16. URL EXTRACTION")

text = "Visit https://example.com or http://test.org/page"

pattern = r"https?://[^\s]+"

print("URLs:", re.findall(pattern, text))


# --------------------------------------------------
# 17. HASHTAGS AND MENTIONS
# --------------------------------------------------
print("\n17. SOCIAL MEDIA TEXT")

text = "Hello @ali! Learning #NLP and #Python."

print("Hashtags:", re.findall(r"#\w+", text))
print("Mentions:", re.findall(r"@\w+", text))


# --------------------------------------------------
# 18. REMOVE EXTRA SPACES
# --------------------------------------------------
print("\n18. NORMALIZE WHITESPACE")

text = "  NLP    is   very \n useful.  "

clean_text = re.sub(r"\s+", " ", text).strip()

print("Clean text:", clean_text)


# --------------------------------------------------
# 19. BASIC NLP TEXT CLEANING
# --------------------------------------------------
print("\n19. NLP TEXT CLEANING")

text = "Hello!!! I am learning NLP in 2026. #AI"

# Remove URLs
text = re.sub(r"https?://\S+|www\.\S+", "", text)

# Remove digits
text = re.sub(r"\d+", "", text)

# Remove punctuation and symbols
text = re.sub(r"[^\w\s]", "", text)

# Normalize whitespace
text = re.sub(r"\s+", " ", text).strip()

print("Cleaned text:", text)


# --------------------------------------------------
# 20. EXTRACT WORDS STARTING WITH A LETTER
# --------------------------------------------------
print("\n20. WORD PATTERNS")

text = "Apple banana Orange mango"

print("Words starting with A:",
      re.findall(r"\bA\w*", text, flags=re.IGNORECASE))

print("Words ending with e:",
      re.findall(r"\b\w*e\b", text, flags=re.IGNORECASE))


# --------------------------------------------------
# 21. FINDITER: MATCH OBJECTS
# --------------------------------------------------
print("\n21. FINDITER")

text = "Order 123, Order 456"

for match in re.finditer(r"\d+", text):
    print("Value:", match.group(), "Position:", match.span())


# --------------------------------------------------
# 22. SUB WITH A FUNCTION
# --------------------------------------------------
print("\n22. REPLACE USING A FUNCTION")

text = "Price: 100, Tax: 20"

result = re.sub(
    r"\d+",
    lambda match: str(int(match.group()) * 2),
    text
)

print("Doubled numbers:", result)


# --------------------------------------------------
# 23. COMPILED REGEX
# --------------------------------------------------
print("\n23. COMPILE PATTERN")

pattern = re.compile(r"\d+")

text = "There are 10 books and 20 pens."

print("Numbers:", pattern.findall(text))


# --------------------------------------------------
# 24. GREEDY VS NON-GREEDY MATCHING
# --------------------------------------------------
print("\n24. GREEDY VS NON-GREEDY")

text = "<p>Hello</p><p>World</p>"

print("Greedy:", re.findall(r"<p>.*</p>", text))
print("Non-greedy:", re.findall(r"<p>.*?</p>", text))


# --------------------------------------------------
# 25. NLP: EXTRACT SENTENCES WITH A SIMPLE PATTERN
# --------------------------------------------------
print("\n25. BASIC SENTENCE SPLITTING")

text = "NLP is useful. Regex is powerful! Python is easy?"

sentences = re.split(r"(?<=[.!?])\s+", text)

print("Sentences:", sentences)


# --------------------------------------------------
# END
# --------------------------------------------------
print("\n" + "=" * 60)
print("ALL REGEX EXAMPLES COMPLETED")
print("=" * 60)

```
