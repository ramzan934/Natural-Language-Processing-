NLU

```python
def understand(text):
    text = text.lower()

    if "weather" in text:
        return {
            "intent": "check_weather"
        }

    if "play music" in text:
        return {
            "intent": "play_music"
        }

    return {
        "intent": "unknown"
    }


print(understand("Can you check the weather?"))
print(understand("Please play music"))
```

Tokenization

```python
nltk.download('punkt')
nltk.download('punkt_tab')
from nltk.tokenize import word_tokenize
from nltk.tokenize import sent_tokenize
```

```python
text = "Natural Language Processing is very interesting."

tokens = word_tokenize(text)

print(tokens)
```

```python
text = "I am learning NLP. It is very useful. I enjoy it."

sentences = sent_tokenize(text)

print(sentences)
```

POS Taggs

```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')
```

```python
from nltk.tokenize import word_tokenize
from nltk import pos_tag

text = "The students are learning Natural Language Processing."

# Step 1: Tokenize the text
tokens = word_tokenize(text)

# Step 2: Perform POS tagging
tags = pos_tag(tokens)

# Step 3: Display the result
print(tags)
```
